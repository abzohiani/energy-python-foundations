```python
# Exercise 1: Convert Megawatts to Kilowatts
def mw_to_kw(power_mw):
    return power_mw * 1000
```

```python
mw_to_kw(35082) # 35082 represents the peak electricity demand in MW on 1/1/2026 
```

```python
# Exercise 2: Calculate Energy (MWh) from Power (MW) and Time (hours)
def calculate_energy(power_mw, time_hours):
    return power_mw * time_hours
```

```python
calculate_energy(35082, 0.5) 
# The calculated output represents the energy in MWh used during the settlement period 35, 
# between 1700 and 1730, which was the time of peak demand on 1/1/2026

```

```python
calculate_energy(21047, 0.5) 
# The calculated output represents the energy in MWh used during the settlement period 12, 
# between 0530 and 0600, which was the time of minimum demand on 1/1/2026
```

```python
# Exercise 3: Calculate Percentage Change 
def percentage_change(old_value, new_value):
    return ((new_value - old_value) / old_value) * 100
```

```python
percentage_change(21047, 35082) 
# The calculated output represents the percentage increase in energy demand during the day
# of 1/1/2026 from 0530 to 1730 (settlement periods 12 to 35)
```

```python
# Exercise 4: Convert List of Half-Hourly Power Values (MW) into Energy Values (MWh)
def half_hourly_to_mwh(power_values):
    total_energy = 0

    for power in power_values:
        total_energy += power * 0.5

    return total_energy
```

```python
demand = [30357, 30466, 31184, 32188, 33178, 34312]

energy = half_hourly_to_mwh(demand)

print(energy)
# This code calculates the energy used in GB in the three hours before peak demand 
# during the day of 1/1/2026 from 1400 to 1700 (settlement periods 29 to 34)
```

```python
# Exercise 5: Calculate Weighted Average Carbon Intensity of the GB Electricity Mix in a Given Period
def calc_co2_emissions(generation, carbon_intensity):
    if len(generation) != len(carbon_intensity):
        raise ValueError("Both lists must have the same length")
    #gen_kWh = mw_to_kw(generation)
    total_gen = sum(generation)
    emissions_grams = sum(gen * carbon for gen, carbon in zip(generation, carbon_intensity))/total_gen
    return emissions_grams

```

```python
demand = [30357, 30466, 31184, 32188, 33178, 34312] 
# Same list of demand during the day of 1/1/2026 from 1400 to 1700 
# (settlement periods 29 to 34)
co2 = [78, 81, 81, 86, 85, 88]
# Corresponding list of carbon intensity values in gCO2/kWh for the same settlement periods
emissions = calc_co2_emissions(demand, co2)
print(emissions)
```
