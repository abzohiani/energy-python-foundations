```python
print(df[["SETTLEMENT_DATE", "SETTLEMENT_PERIOD"]].head()) 
# Settlement date is a datetime dtype. Output from question 2 tells me more data cleaning is required.
# Two problems arise:
# 1. "SETTLEMENT_DATE" returns the time at 00:00:00 for both minimum and maximum demand. 
# In the case of maximum demand this is nonsensical. I need to remove the time from the datetime object.
# 2. A new column, "SETTLEMENT_TIME" in the 24-hour clock needs to be created from "SETTLEMENT_PERIOD".
# Clock changes in March and October need to be accounted for.
```

```python
# Hardcoded 2025 Clock Change Logic - Solving problem 2
def calculate_2025_time(row):
    # Convert date to string format for easy matching
    date_str = row["SETTLEMENT_DATE"].strftime("%Y-%m-%d")
    period = row["SETTLEMENT_PERIOD"]

    # --- SPRING CLOCK CHANGE: 30 March 2025 (46 Periods) ---
    if date_str == "2025-03-30":
        if period > 2:
            # Skip the lost hour (add 2 extra periods worth of minutes)
            minutes_elapsed = (period + 1) * 30
        else:
            minutes_elapsed = (period - 1) * 30

    # --- AUTUMN CLOCK CHANGE: 26 October 2025 (50 Periods) ---
    elif date_str == "2025-10-26":
        if period > 6:
            # Repeat the hour (subtract 2 periods worth of minutes)
            minutes_elapsed = (period - 3) * 30
        else:
            minutes_elapsed = (period - 1) * 30

    # --- STANDARD DAY (48 Periods) ---
    else:
        minutes_elapsed = (period - 1) * 30

    # Calculate hours and minutes mathematically
    hours = (minutes_elapsed // 60) % 24
    minutes = minutes_elapsed % 60

    return f"{hours:02d}:{minutes:02d}"


# Apply the 2025 function using lambda
df["SETTLEMENT_TIME"] = df.apply(lambda row: calculate_2025_time(row), axis=1)

# Print sample dataframe with affected days and times to check the function and one non-affected day as control variable
target_dates = pd.to_datetime(["2025-03-30", "2025-10-26", "2025-10-27"])

print(
    df.loc[
       ((df["SETTLEMENT_DATE"].isin(target_dates)) & (
           df["SETTLEMENT_PERIOD"] <= 7)),
        ["SETTLEMENT_DATE", "SETTLEMENT_TIME", "SETTLEMENT_PERIOD"],
    ]
)


```

```python
# Re-answer Question 2: Find minimum and maximum demand, and when they occurred in 2025

min_demand = df.loc[df["ENGLAND_WALES_DEMAND"].idxmin(), ["SETTLEMENT_DATE", "SETTLEMENT_TIME", "SETTLEMENT_PERIOD", "ENGLAND_WALES_DEMAND"]]
print(min_demand)

max_demand = df.loc[df["ENGLAND_WALES_DEMAND"].idxmax(), ["SETTLEMENT_DATE", "SETTLEMENT_TIME", "SETTLEMENT_PERIOD", "ENGLAND_WALES_DEMAND"]]
print(max_demand)
```

The minimum electricity demand in GB in 2025 was 12,040 MW, which occurred on 25/5/2025 at 14:30. This was a Sunday in late spring. Surprisingly the lowest demand all year wasn't during night hours.
Maximum GB demand for 2025 was 41,835 MW. This occurred on 9/1/2025 at 17:00, a weekday during a dark, cold winter.

```python
# Question 3 — Find average, minimum and maximum demand by month

import calendar

# Get a clean list of all 12 month names ['January', 'February', ...]
ordered_months = list(calendar.month_name)[1:]

# Extract month name and convert it into a ordered Categorical column
df["month_name"] = pd.Categorical(
    df["SETTLEMENT_DATE"].dt.month_name(),
    categories=ordered_months,
    ordered=True,
)

# Now, any groupby or sort operation will naturally stay in perfect calendar order!
monthly_demand_summary = (
    df.groupby("month_name", observed=False)["ENGLAND_WALES_DEMAND"]
    .agg(["mean", "min", "max"])
    .reset_index()
)

# Round the mean column to the nearest integer
monthly_demand_summary["mean"] = monthly_demand_summary["mean"].round(0).astype(int)

print(monthly_demand_summary)

```

This table shows demand peaks in winter and is lowest in summer, as expected. The month of minimum demand
is May, technically late spring, so this is a surprising result.

```python
# Question 4: What are the highest-demand periods of 2025?

top_demand = df.sort_values("ENGLAND_WALES_DEMAND", ascending=False)

# What day of week correspond to the highest demand days?
# Create a new column with the day names
df["SETTLEMENT_DAY"] = df["SETTLEMENT_DATE"].dt.day_name()


print(top_demand[['SETTLEMENT_DAY', 'SETTLEMENT_DATE', 'SETTLEMENT_TIME', 'ENGLAND_WALES_DEMAND']].head(10))
```

The above table shows that the top 10 observations of highest electricity demand by settlement period occur 
in a three-day window in early January, from Tuesday 7/1/2025 to Thursday 9/1/2025. Settlement times for highest demand 
occur between 16:30 and 18:30 corresponding to usual work finish time, when there will be a surge in household heating and 
lighting demand as people arrive from work to a cold dark home in winter.
